```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
customer = pd.read_csv("Mall_Customers.csv")
customer.head()
```

```python
customer["Gender"] = customer["Gender"].map({"Male": 1, "Female": 0})

#Map the Male and female gender to 1 and 0 to avoid value errors
```

```python
customer["Gender"].unique
```

```python
customer.describe().round(1)
```

```python
plt.figure(figsize=(7,5))
plt.scatter(customer["Spending Score (1-100)"], customer["Annual Income (k$)"], alpha=0.7)
```

```python
x = customer[["Annual Income (k$)", "Spending Score (1-100)" ]]

scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
x_scaled[:3].round(2)
```

```python
Kmeans_cluster = KMeans(n_clusters=3, random_state=42, n_init=10)
label = Kmeans_cluster.fit_predict(x_scaled)

plt.figure(figsize=(7,5))
plt.scatter(customer["Annual Income (k$)"], customer["Spending Score (1-100)"], c=label, cmap="viridis", alpha=0.7)
plt.xlabel("Annual Income")
plt.ylabel("Spending score")
plt.title("K-means with k =3 (is this the best choice)")
plt.grid(alpha=0.3)
plt.show()
```

```python
k_value = range(1, 11)
inertias = []
for k in k_value:
    model = KMeans(n_clusters=k, random_state=42, n_init=10).fit(x_scaled)
    inertias.append(model.inertia_)

plt.figure(figsize=(7,4))
plt.plot(k_value, inertias, marker="o")
plt.xlabel("Number of clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow method")
plt.grid(alpha=0.3)
plt.show()
```

```python
final = 5
kmeans = KMeans(n_clusters=final, random_state=42, n_init=10)
customer["cluster"] = kmeans.fit_predict(x_scaled)

centres = scaler.inverse_transform(kmeans.cluster_centers_)

plt.figure(figsize=(7,5))
plt.scatter(customer["Annual Income (k$)"], customer["Spending Score (1-100)"], c=customer["cluster"], cmap="viridis", alpha=0.7)

plt.scatter(centres[:, 0], centres[:, 1], color="red", marker="X", s=200, label="centroids")
plt.xlabel("Annual Income")
plt.ylabel("Spending score")
plt.title(f"Kmeans with k = {final}")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
profile = customer.groupby("cluster").agg(
    customer=("cluster", "size"),
    avg_income =("Annual Income (k$)", "mean"),
    aveg_spending=("Spending Score (1-100)", "mean")
).round(1)
profile
```

```python
new_customer = pd.DataFrame({"Annual Income (k$)": [20], "Spending Score (1-100)": [10]})
segment = kmeans.predict(scaler.transform(new_customer))[0]
print("This customer belongs to cluster:", segment)
```
